## ravel()
#### ->this can conert any array into a 1D array
#### ->it is view of an array i.e.the changes made in view will be reflected in original array

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
arr
```

```python
reshape2 = arr.reshape(3, 2)
reshape2
```

```python
ravel = arr.ravel() #it flattens the array
ravel
```

```python
ravel[0] = 100 #it will change the value of the original array as well because ravel returns a view of the original array
ravel
```

```python
reshape2 #100 is the first element of the original array which is changed to 100 because ravel returns a view of the original array
```
