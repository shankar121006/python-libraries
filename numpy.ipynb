### np.zeros(X)

```python
import numpy as np
```

```python
ar = np.zeros(3)
print(ar)
print(ar.dtype)
print(ar.ndim)
```

```python
ar2 = np.zeros((3, 4))
print(ar2)
print(ar2.dtype)
print(ar2.ndim)
```

```python
ar3 = np.zeros((3, 4, 5))
print(ar3)
print(ar3.dtype)
print(ar3.ndim)
```

### np.ones(x)

```python
import numpy as np
```

```python
ar = np.ones(3)
print(ar)
print(ar.dtype)
print(ar.ndim)
```

```python
ar2 = np.ones((3, 4))
print(ar2)
print(ar2.dtype)
print(ar2.ndim)
```

### np.empty(x)
#### --the elements inside the array can be anything its also called garbage values

```python
import numpy as np
```

```python
ar = np.empty(3)
print(ar)
print(ar.dtype)
print(ar.ndim)
```

```python
ar2 = np.empty((3, 5))
print(ar2)
print(ar2.dtype)
print(ar2.ndim)
```

```python
# meaningful values, but the shape and type are not correct
ar3 = np.empty((3, 3, 4))
print(ar3)
print(ar3.dtype)
print(ar3.ndim)
```

```python
#meaningless values and the shape and type are correct
ar3 = np.empty((2, 3, 4))
print(ar3)
print(ar3.dtype)
print(ar3.ndim)
```

### np.eye()

## np.diag(x)
#### -it will create a square matrix with give elemets as diagonals

```python
import numpy as np
```

```python
ar = np.diag([1, 2, 3, 4, 5])
print(ar)
print(ar.dtype)
print(ar.ndim)
```

```python
ar = np.diag([1, 2, 3])
print(ar)
print(ar.dtype)
print(ar.ndim)
```

## np.arange(start:end:step)
### -same as range function

```python
import numpy as np
```

```python
ar = np.arange(10)
print(ar)
print(ar.dtype)
print(ar.ndim)
```

```python
ar = np.arange(0,10,2)
print(ar)
print(ar.dtype)
print(ar.ndim)
```
